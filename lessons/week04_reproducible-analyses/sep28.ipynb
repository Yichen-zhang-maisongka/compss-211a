```python
import pandas as pd
requests = pd.read_csv("data/sf311_requests.csv")
```

```python
requests.head()
```

```python
original_report = requests.groupby("district")["hours_to_close"].agg(
    requests="size",
    with_time="count",
    median_hours="median"
)

original_report
```

```python
original_report = requests.groupby("district").agg(
    total_requests=pd.NamedAgg(
        column="request_id",
        aggfunc="count"
    )
)

original_report
```

```python

original_report = requests.groupby("district", dropna=False)["hours_to_close"].agg(
    requests="size",
    with_time="count",
    median_hours="median"
)
original_report
```

```python
# 只保留街道和人行道清洁请求
selected = requests.loc[
    requests["category"] == "Street and Sidewalk Cleaning"
].copy()
print("清洁请求数量：", len(selected))

# 调用老师提供的汇总函数
import case_summary
draft_report = case_summary.summarize_requests(selected, group_by="district")

# 有有效处理时长的请求占比（百分数）
draft_report["pct_with_time"] = (
    draft_report["with_time"] / draft_report["requests"]
) * 100
display(draft_report)
```
